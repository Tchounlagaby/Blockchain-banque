### Étape 2 : Mettre à jour `blockchain.js` pour intégrer les signatures numériques

````python
import os

blockchain_path = '/content/blockchain-banque/src/blockchain.js'

code_signatures = """// =============================================================================
// blockchain.js avec Signature Numérique
// =============================================================================
const sha256 = require('sha256');
const EC = require('elliptic').ec;
const ec = new EC('secp256k1');

class Transaction {
    constructor(sender, recipient, amount) {
        this.sender = sender;          // Clé publique de l'expéditeur
        this.recipient = recipient;    // Clé publique du destinataire
        this.amount = amount;
        this.timestamp = Date.now();
        this.signature = null;
    }

    // Calcule le hachage unique de la transaction
    calculateHash() {
        return sha256(this.sender + this.recipient + this.amount + this.timestamp).toString();
    }

    // Signe la transaction avec la clé privée de l'expéditeur
    signTransaction(signingKey) {
        // On s'assure que la clé publique de l'expéditeur correspond à la clé de signature
        if (signingKey.getPublic('hex') !== this.sender) {
            throw new Error("Vous ne pouvez pas signer des transactions pour un autre compte !");
        }

        const hashTx = this.calculateHash();
        const sig = signingKey.sign(hashTx, 'base64');
        this.signature = sig.toDER('hex');
    }

    // Vérifie la validité de la signature
    isValid() {
        // Le bloc Genesis n'a pas d'expéditeur ni de signature
        if (this.sender === null) return true;

        if (!this.signature || this.signature.length === 0) {
            return false;
        }

        const publicKey = ec.keyFromPublic(this.sender, 'hex');
        return publicKey.verify(this.calculateHash(), this.signature);
    }
}

class Blockchain {
    constructor() {
        this.chain = [];
        this.pendingTransactions = [];

        // Création du bloc Genesis
        this.createNewBlock(100, '0', '0');
    }

    createNewBlock(nonce, previousBlockHash, hash) {
        const newBlock = {
            index: this.chain.length + 1,
            timestamp: Date.now(),
            transactions: this.pendingTransactions,
            nonce: nonce,
            hash: hash,
            previousBlockHash: previousBlockHash
        };

        this.pendingTransactions = [];
        this.chain.push(newBlock);

        return newBlock;
    }

    getLastBlock() {
        return this.chain[this.chain.length - 1];
    }

    addTransaction(transaction) {
        if (!transaction.sender || !transaction.recipient) {
            throw new Error('La transaction doit contenir un expéditeur et un destinataire.');
        }

        if (!transaction.isValid()) {
            throw new Error("Impossible d'ajouter une transaction invalide ou falsifiée.");
        }

        this.pendingTransactions.push(transaction);
        return this.getLastBlock()['index'] + 1;
    }

    hashBlock(previousBlockHash, currentBlockData, nonce) {
        const dataAsString = previousBlockHash + nonce.toString() + JSON.stringify(currentBlockData);
        const hash = sha256(dataAsString);
        return hash;
    }

    proofOfWork(previousBlockHash, currentBlockData) {
        let nonce = 0;
        let hash = this.hashBlock(previousBlockHash, currentBlockData, nonce);
        while (hash.substring(0, 4) !== '0000') {
            nonce++;
            hash = this.hashBlock(previousBlockHash, currentBlockData, nonce);
        }
        return nonce;
    }

    isChainValid() {
        for (let i = 1; i < this.chain.length; i++) {
            const currentBlock = this.chain[i];
            const previousBlock = this.chain[i - 1];

            // 1. Vérifier si les transactions à l'intérieur du bloc sont toutes valides
            for (const tx of currentBlock.transactions) {
                // Re-instancier l'objet Transaction pour pouvoir appeler .isValid()
                const tempTx = new Transaction(tx.sender, tx.recipient, tx.amount);
                tempTx.timestamp = tx.timestamp;
                tempTx.signature = tx.signature;
                if (!tempTx.isValid()) {
                    return false;
                }
            }

            // 2. Vérifier l'intégrité des blocs
            const blockData = {
                transactions: currentBlock.transactions,
                index: currentBlock.index
            };
            const recalculatedHash = this.hashBlock(previousBlock.hash, blockData, currentBlock.nonce);

            if (currentBlock.hash !== recalculatedHash) {
                return false;
            }
            if (currentBlock.previousBlockHash !== previousBlock.hash) {
                return false;
            }
        }
        return true;
    }
}

module.exports = { Blockchain, Transaction };
"""

with open(blockchain_path, 'w') as f:
    f.write(code_signatures)

# Recopier à la racine
import shutil
shutil.copy(blockchain_path, '/content/blockchain-banque/blockchain.js')
print("blockchain.js corrigé et mis à jour avec la vérification des signatures cryptographiques !")
````

### Script de simulation : Tentative de recalcul du Proof of Work par un attaquant

Dans ce script, nous allons :
1. Créer une chaîne saine de 3 blocs.
2. Modifier une transaction dans le Bloc #2.
3. Recalculer le Proof of Work (re-miner) pour le Bloc #2 pour tenter de réparer la chaîne.
4. Constater que la chaîne reste invalide tant qu'on ne recalcule pas également le Bloc #3.

````python
test_recalcul_path = '/content/blockchain-banque/test_recalcul.js'

code_test_recalcul = r"""// =============================================================================
// test_recalcul.js
// =============================================================================
const { Blockchain, Transaction } = require('./blockchain');
const EC = require('elliptic').ec;
const ec = new EC('secp256k1');

const maBanque = new Blockchain();
const keyAlice = ec.genKeyPair();
const keyBob = ec.genKeyPair();

// 1. Création du Bloc #2
const tx1 = new Transaction(keyAlice.getPublic('hex'), keyBob.getPublic('hex'), 100);
tx1.signTransaction(keyAlice);
maBanque.addTransaction(tx1);
const block2Data = { transactions: maBanque.pendingTransactions, index: 2 };
const nonce2 = maBanque.proofOfWork('0', block2Data);
const hash2 = maBanque.hashBlock('0', block2Data, nonce2);
maBanque.createNewBlock(nonce2, '0', hash2);

// 2. Création du Bloc #3
const tx2 = new Transaction(keyBob.getPublic('hex'), keyAlice.getPublic('hex'), 50);
tx2.signTransaction(keyBob);
maBanque.addTransaction(tx2);
const block3Data = { transactions: maBanque.pendingTransactions, index: 3 };
const nonce3 = maBanque.proofOfWork(hash2, block3Data);
const hash3 = maBanque.hashBlock(hash2, block3Data, nonce3);
maBanque.createNewBlock(nonce3, hash2, hash3);

console.log("=== 1. État Initial de la Blockchain ===");
console.log("Nombre de blocs :", maBanque.chain.length);
console.log("Validité de la chaîne :", maBanque.isChainValid()); // Doit être true

// 3. L'attaque sur le Bloc #2
console.log("\n=== 2. Attaque et falsification du Bloc #2 ===");
maBanque.chain[1].transactions[0].amount = 999999; // Modification frauduleuse
console.log("Validité après modification simple :", maBanque.isChainValid()); // false

// 4. L'attaquant recalcule le Proof of Work pour le Bloc #2 uniquement
console.log("\n=== 3. Tentative de recalcul du Proof of Work pour le Bloc #2 uniquement ===");
const block2ModifieData = {
    transactions: maBanque.chain[1].transactions,
    index: 2
};

console.log("Re-minage du Bloc #2 en cours...");
const nouveauNonce2 = maBanque.proofOfWork('0', block2ModifieData);
const nouveauHash2 = maBanque.hashBlock('0', block2ModifieData, nouveauNonce2);

// L'attaquant remplace le nonce et le hash dans le Bloc #2
maBanque.chain[1].nonce = nouveauNonce2;
maBanque.chain[1].hash = nouveauHash2;

console.log("Nouveau Hash trouvé pour le Bloc #2 :", nouveauHash2);
console.log("La chaîne est-elle valide maintenant ? :", maBanque.isChainValid());
// Cela reste FALSE car le Bloc #3 pointe toujours vers l'ancien hash2 !

// 5. L'attaquant doit corriger le lien et re-miner le Bloc #3
console.log("\n=== 4. Correction du lien et re-minage du Bloc #3 ===");
maBanque.chain[2].previousBlockHash = nouveauHash2;

const block3ModifieData = {
    transactions: maBanque.chain[2].transactions,
    index: 3
};

console.log("Re-minage du Bloc #3 en cours...");
const nouveauNonce3 = maBanque.proofOfWork(nouveauHash2, block3ModifieData);
const nouveauHash3 = maBanque.hashBlock(nouveauHash2, block3ModifieData, nouveauNonce3);

maBanque.chain[2].nonce = nouveauNonce3;
maBanque.chain[2].hash = nouveauHash3;

console.log("La chaîne est-elle ENFIN valide ? :", maBanque.isChainValid()); // True !
"""

with open(test_recalcul_path, 'w') as f:
    f.write(code_test_recalcul)

print("Fichier test_recalcul.js créé !")
````

### 🛡️ Synthèse des Mécanismes de Sécurité pour le Mémoire

Cette section présente un résumé conceptuel et technique des barrières de sécurité interconnectées qui garantissent l'immutabilité et l'intégrité de notre architecture blockchain bancaire.

```
+---------------------------------------------------------------------------------+
|                                 ESPACE TRANSACTIONNEL                           |
|  [ Transaction ] --( Clé Privée Alice )--> [ Signature Cryptographique ECDSA ]   |
+---------------------------------------------------------------------------------+
                                         |
                                         v (Validation de la signature par le réseau)
+---------------------------------------------------------------------------------+
|                                 ESPACE CONSENSUS                                |
|  [ Bloc N-1 (Hash) ] <=== (Lien) === [ Bloc N (Nonce + Data + PrevHash) ]       |
|                                                |                                |
|                                                v                                |
|                                    [ Proof of Work (SHA-256) ]                  |
+---------------------------------------------------------------------------------+
```

#### 1. L'Intégrité des Données par Signature Numérique (ECDSA)
*   **Principe** : Chaque transaction initiée par un utilisateur (ex: Alice) doit être signée à l'aide de sa clé privée via l'algorithme de signature numérique à courbe elliptique (**ECDSA**, courbe *secp256k1*). La clé publique d'Alice sert d'adresse de compte.
*   **Sécurité** : Si un attaquant tente d'intercepter la transaction et de modifier son contenu (comme le montant ou le destinataire), le hachage de la transaction change. La signature originale devient alors mathématiquement incompatible avec le nouveau hachage. Sans la clé privée d'Alice, le faussaire est dans l'impossibilité de régénérer une signature valide, et le réseau rejette immédiatement la transaction lors de sa soumission.

#### 2. L'Immutabilité Structurelle par Hachage Récursif (Chaînage de Blocs)
*   **Principe** : Chaque bloc valide contient le hachage cryptographique du bloc précédent (`previousBlockHash`).
*   **Sécurité** : Ce chaînage crée une dépendance temporelle stricte. Si la moindre donnée historique à l'intérieur d'un bloc antérieur est altérée, le hachage de ce bloc est complètement modifié (effet d'avalanche). Par conséquent, la référence contenue dans le bloc suivant devient obsolète, brisant instantanément la chaîne de confiance à partir de ce point.

#### 3. Le Consensus et la Résistance aux Attaques via Proof of Work (PoW)
*   **Principe** : La validation d'un nouveau bloc requiert la résolution d'un problème mathématique complexe (le minage). Le protocole impose que le hachage du bloc commence par un nombre défini de zéros (la difficulté, ici réglée sur `0000`), ce qui exige de trouver une valeur unique appelée `nonce` par force brute.
*   **Sécurité** : Si un pirate informatique souhaite falsifier une transaction passée, il ne lui suffit pas de modifier la donnée. Il doit :
    1. Retrouver un nouveau `nonce` valide pour réhacher le bloc falsifié.
    2. Mettre à jour le pointeur `previousBlockHash` du bloc suivant et recalculer également le Proof of Work de ce dernier.
    3. Répéter cette opération de re-minage pour **chaque bloc ultérieur** jusqu'au sommet actuel de la chaîne.

Pour que sa chaîne falsifiée soit acceptée par les autres nœuds du système (règle de la chaîne la plus longue), l'attaquant devrait être capable de recalculer ces équations plus vite que l'ensemble cumulé des autres participants du réseau. Cela requiert un monopole de la puissance de calcul du réseau (la barrière théorique de l'**attaque des 51 %**), rendant l'altération de l'historique économiquement et techniquement irréalisable.

### 🔑 L'Importance Cruciale du Hachage Cryptographique dans l'Intégrité de la Chaîne

Le hachage cryptographique constitue la pierre angulaire technologique de toute blockchain. Sans cette fonction mathématique unilatérale, il serait impossible d'assurer la confiance décentralisée et de l'associer à une structure de données inviolable.

#### 1. Qu'est-ce qu'une Fonction de Hachage Cryptographique (SHA-256) ?
Une fonction de hachage prend n'importe quelle quantité de données en entrée (un texte, une image, ou l'intégralité d'un registre de transactions) et la convertit en une empreinte numérique de taille fixe (pour le **SHA-256**, une chaîne de 64 caractères hexadécimaux représentant 256 bits).

Pour être qualifiée de cryptographique, cette fonction doit respecter trois propriétés fondamentales :
*   **Unidirectionnalité (One-Way Function)** : Il est mathématiquement impossible de retrouver les données d'origine à partir du hachage obtenu.
*   **Résistance aux collisions** : Deux ensembles de données d'entrée différents ne doivent pas produire le même hachage.
*   **L'effet d'avalanche (Avalanche Effect)** : La moindre modification d'un seul bit dans les données d'entrée (par exemple, remplacer un point par une virgule, ou modifier un montant de `10.00` en `10.01`) génère un hachage de sortie radicalement et imprévisiblement différent.

```
Données saines     : "Alice paie 10€ à Bob"  ===> SHA-256 ===> 3a5f82c... (Signature unique)
Données altérées   : "Alice paie 100€ à Bob" ===> SHA-256 ===> 9d2e1b4... (Totalement différent)
```

#### 2. Le Rôle du Hachage dans le Chaînage Temporel (La Cohésion d'Historique)
Dans une blockchain, chaque bloc possède un en-tête contenant la clé cryptographique du bloc qui le précède immédiatement (`previousBlockHash`).

Ce lien cryptographique engendre une réaction en chaîne :
1. Si un pirate tente de modifier rétroactivement une transaction dans le **Bloc #10**.
2. L'empreinte globale (le hachage) du **Bloc #10** change instantanément en vertu de l'effet d'avalanche.
3. Le **Bloc #11**, qui stockait le hachage d'origine du Bloc #10, se retrouve à pointer vers une valeur invalide.
4. Par conséquent, la signature globale du **Bloc #11** change également, ce qui invalide le **Bloc #12**, et ainsi de suite.

Le hachage transforme ainsi un registre distribué en un **château de cartes virtuel** : toucher à une seule base détruit instantanément la validité de toute la structure qui repose dessus.

#### 3. Le Hachage comme Outil de Consensus (Preuve de Travail)
Le hachage n'assure pas seulement l'immutabilité passive de l'historique ; il est aussi le moteur de la création de la confiance. Le mécanisme de Proof of Work consiste à contraindre les machines du réseau à chercher de manière répétitive une variable aléatoire (`nonce`) qui, combinée aux transactions, génère un hachage global du bloc inférieur à une certaine cible de difficulté (ex: commençant par `0000`).

C'est la difficulté de calcul de ce hachage combinée à la facilité déconcertante de sa vérification par le reste du réseau qui permet d'atteindre un consensus décentralisé fiable, sans organe central de contrôle.

### ⚖️ Analyse Comparative : Hachage Cryptographique vs Signature Numérique

Bien que le **hachage** et la **signature numérique** reposent tous deux sur des principes de cryptographie, ils remplissent des fonctions de sécurité distinctes et hautement complémentaires au sein d'une blockchain.

Le tableau ci-dessous résume et compare ces deux piliers :

| Caractéristique | Le Hachage Cryptographique (SHA-256) | La Signature Numérique (ECDSA) |
| :--- | :--- | :--- |
| **Objectif Principal** | Garantir l'**intégrité** globale et l'**immutabilité** de la structure (la chaîne). | Garantir l'**authenticité** et le **non-répudiation** d'une transaction (l'expéditeur). |
| **Type de Cryptographie** | Cryptographie symétrique / sans clé (fonction de hachage à sens unique). | Cryptographie asymétrique (système à clé privée / clé publique). |
| **Donnée d'Entrée** | Le bloc entier (transactions + métadonnées + hachage précédent + nonce). | Le hachage de la transaction unique + la clé privée de l'expéditeur. |
| **Vérification** | Rapide : recalcul du hachage par n'importe quel nœud pour vérification d'égalité. | Vérification mathématique de la signature à l'aide de la clé publique de l'expéditeur. |
| **Protection contre...** | La modification de l'historique de la chaîne (altération de données passées). | L'usurpation d'identité, l'injection de transactions fictives et le vol de fonds. |

#### 1. Complémentarité des deux mécanismes

*   **Le Hachage protège la structure (le contenant)** : Il garantit qu'aucun bloc déjà inscrit dans l'historique ne puisse être modifié sans que cela ne brise la chaîne de hachage de manière visible.
*   **La Signature protège le contenu (les transactions)** : Elle garantit que seules les personnes possédant la clé privée d'un compte peuvent initier ou modifier une transaction de ce même compte.

#### 2. Exemple de scénario d'attaque : L'effet de synergie

Imaginons un attaquant tentant de modifier le montant d'une transaction passée pour s'attribuer des fonds :

1.  **S'il n'y avait que le hachage** : L'attaquant possédant une puissance de calcul immense pourrait recalculer tous les Proof of Work (re-minage) des blocs suivants pour rendre la chaîne à nouveau valide. La fraude passerait inaperçue.
2.  **S'il n'y avait que la signature** : La transaction falsifiée serait immédiatement rejetée, car l'attaquant ne peut pas signer le nouveau montant sans la clé privée de la victime. Cependant, sans le chaînage par hachage des blocs, un attaquant pourrait réorganiser l'ordre des transactions légitimes (attaque par double dépense) sans altérer les signatures.

**Conclusion pour le mémoire** : La sécurité d'une blockchain bancaire est un système à double verrou. Le hachage scelle le registre de manière temporelle, tandis que la signature numérique verrouille la légitimité individuelle de chaque mouvement de fonds. Ensemble, ils éliminent le besoin d'un tiers de confiance centralisé.
